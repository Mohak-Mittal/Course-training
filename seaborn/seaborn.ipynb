1. Seaborn-> Seaborn is a Python data visualization library built on top of Matplotlib. It helps us create attractive and informative statistical graphs with relatively less code.

It helps us understand patterns, relationships, distributions, trends, and differences in datasets.


Why do we use Seaborn?

Suppose we have a dataset containing students' study hours, attendance percentages, and examination marks.

Using Seaborn, we can investigate:
- Whether study hours are associated with examination marks.
- How marks are distributed among students.
- Whether attendance differs across performance groups.
- Whether the dataset contains potential outliers.
- How different variables are related to one another.

```python
# Let's load the tips dataset.

import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

df = sns.load_dataset("flights")

print(df.head(2))
print(df.shape)
print(df.columns)
print(df.info())
print(df.describe())
```

```python
sns.scatterplot(
    data=df,
    x="year",
    y="passengers"
)

plt.title("Total Bill vs Tip")
plt.xlabel("Total Bill")
plt.ylabel("Tip Amount")

plt.show()
```

Seaborn Distribution Plots

1. What is a distribution plot?
A distribution describes how values of a variable are spread across a dataset.

For example, suppose we have the examination marks of 100 students. We may want to know:
- Which marks occur most frequently?
- Are most students scoring between 60 and 80?
- Is the distribution symmetric or skewed?
- Are there unusually high or low values?
- How does one group's distribution compare with another's?
Distribution plots help answer these questions visually.

Types of distribution plots in Seaborn









1. histplot()

Displays the frequency or density of observations using bins. Useful for understanding how numerical values are distributed.












2. kdeplot()

Displays a smoothed estimate of the data distribution. Useful for examining its overall shape.












3. displot()

A figure-level function that creates distribution visualizations, including histograms, KDE plots, and ECDF plots.












4. ecdfplot()

Shows the cumulative proportion of observations less than or equal to each value.












5. rugplot()

Displays individual observations as small marks along an axis.

histplot() — Histogram

```python
# A histogram represents the distribution of a numerical variable by grouping values into intervals called bins.

# For example, consider these marks:

# 42, 48, 51, 55, 58, 62, 65, 68, 72, 75, 78, 82, 85, 91

# Instead of displaying every score separately, we group the scores into intervals, such as:
# Marks interval	Frequency
# 40-49	                 2
# 50-59	                 3
# 60-69	                 3
# 70-79	                 3
# 80-89                  2
# 90-99	                 1

# The frequency tells us how many observations fall within each interval.
```

```python
# Basic syntax
# sns.histplot(data=None, x=None, bins="auto", kde=False)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset("tips")

sns.histplot(data=df, x="total_bill",bins = 3)

plt.title("Distribution of Total Bills")
plt.xlabel("Total Bill Amount")
plt.ylabel("Frequency")

plt.show()
```

```python
# bins->A bin is an interval used to group numerical observations.

# sns.histplot(data=df, x="total_bill", bins=5)
# plt.title("Histogram with 5 Bins")
# plt.show()

# sns.histplot(data=df, x="total_bill", bins=20)
# plt.title("Histogram with 20 Bins")
# plt.show()

sns.histplot(data=df, x="total_bill", bins="auto",kde = True)
plt.show()

# sns.histplot(data=df, x="total_bill", binwidth=5)
# plt.show()
```

```python
# stat-> By default, histplot() displays counts. The stat parameter lets us display other statistics.
```

```python
sns.histplot(data=df, x="total_bill", stat="count")
plt.title("Count Histogram")
plt.show()
```

```python
sns.histplot(data=df, x="total_bill", stat="density")
plt.title("Density Histogram")
plt.show()
```

```python
# "count"	       Number of observations in each bin
# "frequency"	   Count divided by bin width
# "probability"	   Each bin's proportion of observations
# "percent"	       Each bin's percentage of observations
# "density"	       Normalized density; total histogram area equals 1
```

```python
# KDE curve to a histogram
# A KDE curve provides a smoothed estimate of the distribution.

sns.histplot(
    data=df,
    x="total_bill",
    bins=15,
    kde=True
)

plt.title("Histogram with KDE")
plt.show()
```

kdeplot() — Kernel Density Estimation

```python
# What is KDE?
# Kernel Density Estimation estimates the underlying probability density of a continuous variable.

# Instead of grouping observations into bins, KDE places a smooth contribution around each observation 
# and combines those contributions into a curve.

# It is useful for understanding:
# - Where observations are concentrated.
# - Whether a distribution has one or several peaks.
# - How two groups' distributions differ.
# - Whether a distribution is approximately symmetric or skewed.
```

```python
sns.kdeplot(data=df, x="total_bill")

plt.title("Density Distribution of Total Bills")
plt.xlabel("Total Bill")
plt.ylabel("Density")

plt.show()
```

Changing KDE smoothing

```python
sns.kdeplot(
    data=df,
    x="total_bill",
    bw_adjust=0.2
)
plt.title("KDE with Less Smoothing")
plt.show()
```

```python
sns.kdeplot(
    data=df,
    x="total_bill",
    bw_adjust=2
)
plt.title("KDE with More Smoothing")
plt.show()
```

```python
# A smaller bandwidth adjustment can reveal more local variation.

# A larger bandwidth adjustment creates a smoother curve and may hide smaller peaks.
```


Filling the KDE curve

```python
sns.kdeplot(
    data=df,
    x="total_bill",
    fill=True
)

plt.title("Filled KDE Plot")
plt.show()
```

displot() — Figure-level distribution plotting

```python
# What makes displot() different?
# displot() is a figure-level function. 
# It provides a convenient way to create distribution plots and organize them into multiple panels.

# It can create histograms, KDE plots, and empirical cumulative distribution plots.
```

```python
sns.displot(
    data=df,
    x="total_bill",
    kind="hist",
    bins=15
)

plt.show()
```

```python
sns.displot(
    data=df,
    x="total_bill",
    kind="kde",
    fill=True
)

plt.show()
```

```python
sns.displot(
    data=df,
    x="total_bill",
    hue="time",
    kind="hist",
    multiple="fill"
)

plt.show()
# Compare groups


# Other multiple options include:
# - "layer".
# - "dodge".
# - "stack".
# - "fill".
```

ecdfplot() — Empirical Cumulative Distribution

```python
# What is an ECDF?
# An empirical cumulative distribution function (ECDF) shows the proportion of observed values
#  that are less than or equal to a given value.

# For example, if 80% of students score 70 or below, the ECDF at 70 is approximately 0.80.
```

```python
sns.ecdfplot(
    data=df,
    x="total_bill"
)

plt.title("Cumulative Distribution of Total Bills")
plt.xlabel("Total Bill")
plt.ylabel("Cumulative Proportion")

plt.show()

# How to interpret the graph
# - The X-axis represents bill amounts.
# - The Y-axis represents the cumulative proportion, from 0 to 1.
# - At a selected X-value, the curve tells us what proportion of observations are at or below that value.

# ECDF plots are especially useful when:
# - Comparing distributions without choosing histogram bins.
```

```python
# Comparing two categories
sns.ecdfplot(
    data=df,
    x="total_bill",
    hue="time"
)

plt.title("ECDF of Lunch and Dinner Bills")
plt.show()
```

rugplot() — Individual observations

```python
# What is a rug plot?
# A rug plot represents each observation as a small mark along an axis.
```

```python
sns.rugplot(
    data=df,
    x="total_bill"
)

plt.title("Individual Total Bill Observations")
plt.show()

# Each small mark represents an individual bill amount.

# Where marks are tightly concentrated, 
#many observations have similar values. Where marks are sparse, fewer observations occur.
```

Combining a rug plot with KDE

```python
sns.kdeplot(
    data=df,
    x="total_bill",
    fill=True
)

sns.rugplot(
    data=df,
    x="total_bill"
)

plt.title("KDE with Individual Observations")
plt.show()

# The KDE shows the estimated overall density, while the rug marks show where the actual observations lie.

# Rug plots can become cluttered when a dataset contains many observations, 
# so they are most useful with small or moderately sized datasets.
```

Mini-project: Analyze restaurant bills

```python
import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset("tips")

sns.set_theme(style="whitegrid")

fig, axes = plt.subplots(2, 2, figsize=(12, 8))

sns.histplot(
    data=df,
    x="total_bill",
    bins=15,
    kde=True,
    ax=axes[0, 0]
)
axes[0, 0].set_title("Histogram with KDE")

sns.kdeplot(
    data=df,
    x="total_bill",
    hue="time",
    fill=False,
    ax=axes[0, 1]
)
axes[0, 1].set_title("KDE by Meal Time")

sns.ecdfplot(
    data=df,
    x="total_bill",
    hue="time",
    ax=axes[1, 0]
)
axes[1, 0].set_title("ECDF by Meal Time")

sns.rugplot(
    data=df,
    x="total_bill",
    ax=axes[1, 1]
)
axes[1, 1].set_title("Individual Bill Observations")

plt.tight_layout()
plt.show()
```
