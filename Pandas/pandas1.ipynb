```python
# data cleaning:

import pandas as pd
import numpy as np

data = {
    "Name": ["Ankita","Priya","Rahul"," Akassh"],
    "Age" : [24,np.nan,67,56],
    "Salary": [67000,56000,np.nan, 450000]
}
df = pd.DataFrame(data)
print(df)
```

```python
# isnull():
print(df.isnull().sum())
# true : missing
# .false: present
```

```python
# fillna()
import numpy as np
import pandas as pd
data = {
    "Name": ["Ankita","Priya","Rahul"," Akassh"],
    "Age" : [24,np.nan,67,56],
    "Salary": [67000,56000, np.nan,450000]
}
df = pd.DataFrame(data)
print(df)
# df.fillna(0)

# # # fill Age with Average
df["Age"]= df["Age"].fillna(df["Age"].mean())
print(df["Age"])
df["Salary"]= df["Salary"].fillna(df["Salary"].mean())
print(df["Salary"])
print(df)
```

```python
# dropna():
import numpy as np
import pandas as pd
data = {
    "Name": ["Ankita","Priya","Rahul"," Akassh"],
    "Age" : [24,np.nan,67,56],
    "Salary": [67000,56000, np.nan,450000]
}
df = pd.DataFrame(data)
print(df)
df.dropna(subset= ["Salary"])
print(df)
```
