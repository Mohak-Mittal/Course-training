```python
# 1. Array manipulation kya hota hai?

# Array manipulation ka simple meaning hai array ka shape change karna, data ko arrange karna, ya arrays ko combine aur divide karna.
```

```python
import numpy as np

a=np.array([1,2,3,4,5,6,7])
print(a)
```

```python
# Ab hum is array ko different ways mein change kar sakte hain:

# 1.reshape() — array ka shape change karna.

# 2.flatten() — multidimensional array ko 1D banana.

# 3.ravel() — array ko 1D banana.

# 4.transpose() — rows ko columns aur columns ko rows banana.

# 5.concatenate() — arrays ko jodna.

# 6.split() — array ko parts mein divide karna.

# 7.sort() — elements ko order mein arrange karna.
```

```python
# 1.reshape() shape change krna

import numpy as np 
arr = np.arange(1,13)
print(arr)
matrix = arr.reshape(3,4)
print(matrix)
```

```python
# 2.Flatten()  2D array ko 1D banana
arr = np.array([
    [10,20],
    [30,40]
])
print(arr)
d = arr.flatten()
print(d)
```

```python
# 3.ravel() aaray ko 1D banana

arr = np.array([
    [10,20,30],
    [40,50,60]

])

print(arr)

b = arr.ravel()
print(b)
```

```python
# Ab tum sochoge ki flatten() aur ravel() same hain.
#  Output generally same hota hai, lekin important difference hai.

# flatten()

	

# ravel()

# Always copy banata hai

# Possible ho toh original data ka view deta hai

# Memory alag use karta hai

# Often memory save karta hai

# Original data se independent result

# View hone par changes original ko affect kar sakte hain
```

```python
# 4. transpose()  Rows ko columns bnana 
import numpy as np
a = np.array([
    [1,2,3],
    [4,5,6]
])

print(a.T)
```

```python
# 5.concatenate()  Do arrays ko combine krna 

a = np.array([1,2,3])
b= np.array([4,5,6])

c = np.concatenate((a,b))
print(c)
```

```python
a = np.array([
[1,2],
[3,4]
])
b = np.array([
    [5,6],
    [7,8]
])

c = np.concatenate((a,b), axis = 0)

print(c)
```

```python
a = np.array([
[1,2],
[3,4]
])
b = np.array([
    [5,6],
    [7,8]
])

c = np.concatenate((a,b), axis = 1)

print(c)
```

```python
# axis = 0  means rows 
# axis = 1 means column
```

```python
# 6. split() Array ko parts mein divide krna 

a = np.array([1,2,3,4,5,6])

b = np.split(a,4)

print(b)
```

```python
# 7. sort() Elements ko arrange krna 

a = np.array([50, 10, 40, 20, 30])

b = np.sort(a)

print(b)
```
