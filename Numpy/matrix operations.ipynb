```python
# Scalar Multiplication
# Jab matrix ke har element ko ek single number se multiply karte hain, usse scalar multiplication kehte hain.
# A = [1 2]
    # [3 4]
# 3A=3[[1,2][3,4]]
# 3A = [3,6],[9,12]
import  numpy as np
arr = np.array([[1,2,5],[1,4,7]])
result = 3 * arr
print(result)
```

```python
# Element-wise Multiplication
# NumPy mein A * B ka matlab hai same position ke elements ko multiply karna

arr = np.array([[1,2,5],
                [1,4,7]])

arr1 = np.array([[1,2,5],
                 [1,4,7]])
print(arr*arr1)
```

```python
# Matrix Multiplication
# Matrix multiplication mein ek matrix ki row ko doosri matrix ke column se multiply karke sum karte hain.

# A = [1,2],
#     [3,4]

# B =[5,6],
    # [7,8]

# FIRST ROW WITH FIRST COLUMN
# (1X5)+(2X7)=19

# FIRST ROW WITH SECOND COLUMN
# (1X6)+(2X8)==22

# AB = [19,22],
#   [43,50]
# NUMPY CODE
import numpy as np
a = np.array([[2,3],[5,7]])
b = np.array([[8,7],[9,6]])
result =np.dot(a,b)
print(result)
```

```python
# # WE CAN ALSO USE np.matmul(A,B)
# Operation	         Meaning
# A * B	             Corresponding elements multiply hote hain.
# A @ B	             Proper matrix multiplication hoti hai.
# np.dot(A, B)	     2D arrays ke liye matrix multiplication jaisa result. 
```

```python
# SHAPE RULE
# if matrix A is (m,n) and B is (p,q) 
# in this condition multiplication is possible if n=p

# then result shape is (m,q)

# example = A = (2,3)
        #   B = (3,4)
# RESULT = (2,4)
```

```python
# a = (m,n)
# b = (p,q)
# n==p
# result = (m,n)
```

```python
A = np.array([[1, 2, 3],
              [4, 5, 6]])

B = np.array([[1, 2],
              [3, 4],
              [5, 6]])
# =((1)+6+15),2+8+18
#   (4+15+30),(8+20+36)

print(A.shape)       # (2, 3)
print(B.shape)       # (3, 2)
print(A @ B)
```

```python
# Transpose of a Matrix
# ROWS WILL BECVOME COLUMNS ANS VICE VERSA

# A = [1 2 3]
    # [4 5 6]
# A' = [1 4]
    #   [2 5]
#      [3 6]

A = np.array([[1, 2, 3],
              [4, 5, 6]])

print(A.T)
print("Shape:", A.T.shape)
```

```python
# Determinant of a Matrix

# Determinant ek single numerical value hoti hai jo square matrix se calculate hoti hai.
# A = [1,2],

#     [3,4]

# det(A)=ad-bc
A = np.array([[1, 2],
              [3, 4]])

det = np.linalg.det(A)

print(det)
# np.linalg NumPy ka linear algebra module hai.
# det() determinant calculate karta hai.
# Determinant zero hone par square matrix ka inverse exist nahi karta.
```

```python
# Inverse of a Matrix
# H. Inverse of a Matrix
# Inverse matrix ko A^-1 se represent karte hain.

# if inverse of matrix exists
# AA^-1=A^-1A=I
# AA**{-1}=A**{-1}A=I
# here I is identity matrix

# a = [1 3]
#     [3 2]
A = np.array([[1, 2],
              [3, 4]])
# I = [1 0]
#     [0 1]
# AA^-1=I
#   1     [4  -2]
# -----  *
#  -2  [-3  1]
# [-2 1]
# [1.5 -0.5]

inverse = np.linalg.inv(A)

print(inverse)
```

```python
# Dot Product in NumPy
# Do vectors ke corresponding elements ko multiply karke un sab products ka sum nikalna dot product kehlata hai.
```

```python

# a=[1,2,3]

# b=[4,5,6]
# Calculation:

# a.b=(1*4)+(2*5)+(3*6)
# =4+10+18=32
import numpy as np

a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

result = np.dot(a, b)

print(result)
# a pehla vector hai.
# b doosra vector hai.
# np.dot(a, b) corresponding elements ko multiply karke sum karta hai.
# result ek scalar value hai.

a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print(a * b)
print(np.dot(a, b))
```

```python
# Dot product with 2D matrices

A = np.array([[1, 2],
              [3, 4]])

B = np.array([[5, 6],
              [7, 8]])
# [5+14    6+16
print(np.dot(A, B))
```

```python
# Linear Algebra in NumPy
# Linear Algebra mathematics ki ek branch hai jisme hum 
# vectors, matrices, linear equations aur unke relationships ko study karte hain.

# linear algebra functions  are in np.linalg module of numpy module

import numpy as np

A = np.array([[1, 2],
              [3, 4]])

print(np.linalg.det(A))
print(np.linalg.inv(A))
```

```python
# Identity Matrix
# I = [1 0 0]
    # [0 1 0]
    # [0 0 1]  

import numpy as np

I = np.eye(4)

print(I)
```

```python
# Solving Linear Equations

# 2x+y=8
# x+3y=9
# WE HAVE TO FIND X,Y WHO SATISFY BOTH EQUATIONS

# IN MATRIX FORM THESE EQUATIONS ARE: AX=B

# WHERE= A = [2 1]    B = [8]   X = [x]
#            [1 3]        [9]       [y]

import numpy as np

A = np.array([[2, 1],
              [1, 3]])

B = np.array([8, 9])

solution = np.linalg.solve(A, B)

print(solution)
```

```python
# Eigenvalues and Eigenvectors
# Eigenvector: Ek non-zero vector jo matrix se transform hone par apni direction preserve karta hai.
# Eigenvalue: Woh factor jisse eigenvector ki magnitude scale hoti hai.
# Mathematical relation:
# Av=lambda v

# (A) = matrix
# (v) = eigenvector
#lambda = eigenvalue

import numpy as np

A = np.array([[2, 0],
              [0, 3]])

values, vectors = np.linalg.eig(A)

print("Eigenvalues:")
print(values)

print("Eigenvectors:")
print(vectors)
```
