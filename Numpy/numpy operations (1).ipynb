```python
# Introduction to NumPy

# NumPy stands for Numerical Python. It is a popular Python library used to perform mathematical and numerical operations on data.

# NumPy works with arrays and allows us to perform calculations on many values at the same time.
```

```python
# Import the library into your Python program:

# import numpy as np
```

```python
# Basic mathematical operations

# For the following examples, we will use two arrays:

import numpy as np

a = np.array([10, 20, 30])
b = np.array([2, 4, 5])

# The arrays contain three elements each. NumPy can perform calculations on corresponding elements.
```

```python
# Addition — np.add()

# Definition: Addition combines two values by adding one value to another.

# Syntax:

# np.add(array1, array2)

# Program:

a = np.array([10, 20, 30])
b = np.array([2, 4, 5])

result = np.add(a, b)

print(result)
```

```python
# B. Subtraction — np.subtract()

# Definition: Subtraction finds the difference between corresponding elements of two arrays.

# Syntax:

# np.subtract(array1, array2)

# Program:

a = np.array([10, 20, 30])
b = np.array([2, 4, 5])

result = np.subtract(a, b)

print(result)
```

```python
# C. Multiplication — np.multiply()

# Definition: Multiplication multiplies corresponding elements of two arrays.

# Syntax:

# np.multiply(array1, array2)

# Program:

a = np.array([10, 20, 30])
b = np.array([2, 4, 5])

result = np.multiply(a, b)

print(result)
```

```python
# D. Division — np.divide()

# Definition: Division divides each element of the first array by the corresponding element of the second array.

# Syntax:

# np.divide(array1, array2)

# Program:

a = np.array([10, 20, 30])
b = np.array([2, 4, 5])

result = np.divide(a, b)

print(result)
```

```python
# 4. Advanced mathematical operations
# A. Power — np.power()

# Definition: Power raises each element of an array to a specified exponent.

# For example, \(2^3 = 2 \times 2 \times 2 = 8\).

# Program:

import numpy as np

a = np.array([2, 3, 4])

result = np.power(a, 2)

print(result)
```

```python
# B. Square root — np.sqrt()

# Definition: The square root of a number is a value that, when multiplied by itself, gives the original number.

# Program:

import numpy as np

a = np.array([4, 9, 16, 25])

result = np.sqrt(a)

print(result)
```

```python
# C. Modulus — np.mod()

# Definition: Modulus returns the remainder after dividing one number by another.

# Program:

import numpy as np

a = np.array([10, 15, 20])
b = np.array([3, 4, 6])

result = np.mod(a, b)

print(result)
```

```python
# D. Absolute value — np.abs()

# Definition: Absolute value returns the non-negative magnitude of a number.

# Program:

import numpy as np

a = np.array([-10, -5, 0, 5, 10])

print(np.abs(a))

# Explanation: Negative values become positive, positive values remain unchanged, 
# and zero remains zero.
```

```python
# E. Rounding — np.round()

# Definition: Rounding changes a numerical value to a specified number of decimal places.

import numpy as np

a = np.array([2.345, 4.678, 7.891])

print(np.round(a, 2))
```

```python
# Statistical and aggregate operations

# These functions calculate useful summary values from an array.

# A. Sum — np.sum()

# Adds all the elements of an array.

a = np.array([10, 20, 30, 40])

print(np.sum(a))
```

```python
# B. Mean — np.mean()

# Calculates the arithmetic average.



# Program:

a = np.array([10, 20, 30, 40])

print(np.mean(a))
```

```python
# C. Maximum — np.max()

# Finds the largest value in an array.

a = np.array([10, 50, 30, 20])

print(np.max(a))
```

```python
# D. Minimum — np.min()

# Finds the smallest value in an array.

a = np.array([10, 50, 30, 20])

print(np.min(a))
```

```python
# E. Standard deviation — np.std()

# Standard deviation measures how spread out values are around their mean.

a = np.array([10, 20, 30])

print(np.std(a))
```
