```python
#
# for example:  filter means selecting only those value from array that satisfy or based on  given condition
import numpy as np
arr = np.array([20,30,50,60])
result = np.where(arr >= 50,'pass','fail')
print(result)
```

```python
# comparison used compare the two or more values using  comparison operator like greater than,greater than or equals to , less than or equals to, less than,equals to ,not equals to 

# for example
import numpy as np
arr = np.array([20,30,40,50,60,70,80])
result = np.where(arr > 50,arr,' ' )
print(result)
# less than
result = np.where(arr < 50,arr,' ')
print(result)
# equals to
result = np.where(arr == 50,arr,' ')
print(result)
# not equals to
result = np.where(arr != 50,arr,' ')
print(result)
# greator than equals to
result = np.where(arr >= 50,arr,' ')
print(result)
# less than equals to
result = np.where(arr <= 50,arr,' ')
print(result)
```

```python
# Boolean : it returns True if the condition is satisfy  and if not it return False
# for example

arr =np.array([50,80,60,30])
result = np.where(arr > 50,True,False )
print(result)

```

```python
# lets take an example with boolean operator
# & it returns True when both condition True
arr =np.array([50,80,60,30])
result = np.where((arr > 20) & (arr < 50),True,False)
print(result)
```

```python
# | in this operator it return true when atleast one condition must be True 
# for example
arr =np.array([50,80,60,82])
result = np.where((arr > 85) | (arr < 80),True,False)
print(result)
```

```python
# ~ in this , if the condition is True it return false like vice - versa
# for example
arr =np.array([50,80,60,30])
result = np.where(~( arr < 55),True,False)
print(result)
```
