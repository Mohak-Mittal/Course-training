```python
# Broadcasting: Broadcasting is a feature in NumPy that allows us to perform arithmetic operations on arrays of different but compatible 
# shapes without changing their original shapes manually.

# Why do we use Broadcasting in NumPy?
# To perform operations easily: We can perform calculations on arrays of different compatible shapes.
# To avoid repeating values manually: NumPy automatically broadcasts smaller arrays or values when needed.
# To save time and code: We don't need to resize arrays manually.
# To improve efficiency: Broadcasting allows NumPy to perform calculations efficiently.
```

```python
# Example:

import numpy as np

a = np.array([10, 20, 30])
print(a + 5)


# Here, NumPy automatically adds 5 to every element of the array.
```

```python
# RULES FOR BROADCASTING:

# Dimensions must be equal or one of them must be 1.

import numpy as np

a = np.array([[1, 2, 3]])
b = np.array([[10, 20, 30]])

print(a + b)


# NumPy compares dimensions from right to left.

# import numpy as np

a = np.array([[1, 2, 3],
              [4, 5, 6],
              [4,7,8]])

b = np.array([10, 20, 30])

print(a + b)

# # If dimensions are incompatible, it raises a ValueError.

# import numpy as np

a = np.array([1, 2, 3])
b = np.array([10, 20])

print(a + b)



```

```python
# Broadcasting with Random Numbers

import numpy as np

a = np.random.randint(1, 10, size=(3,1))
b = np.random.randint(1, 10, size=(1,3))

print("Array A:\n", a)
print("Array B:\n", b)

print("Result:\n", a + b)


# ya to same ya to anyone contains 1

# Explanation:

# np.random.randint() generates random integers.

# size=(3,1) creates an array with 3 rows and 1 column.

# size=(1,3) creates an array with 1 row and 3 columns.

# a + b adds the arrays using broadcasting.

# The result has a shape of (3,3).

# The random numbers and output will change each time you run the code.
```

```python
# Definition : Vectorization is a technique in NumPy that allows us to perform operations on entire arrays
#  without using explicit Python loops.
# EXAMPLE:

import numpy as np

a = np.array([1, 2, 3])
print(a + 5)


import numpy as np

a = np.array([2, 3, 4])
print(a ** 2)

# Why do we use it?

# To make code shorter.

# To reduce the need for loops.

# To perform calculations efficiently.
```

```python
# EXAMPLE :

import numpy as np

a = np.random.randint(1, 10, size=5)

print("Original array:", a)
print("After multiplication:", a * 2)
print(type(a))

# Explanation:

# np.random.randint(1, 10, size=5) generates 5 random integers.

# a * 2 multiplies every element by 2.

# NumPy performs the operation on the entire array without a Python loop.
```
