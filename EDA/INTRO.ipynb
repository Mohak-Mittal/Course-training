```python
# MISSING VALUES
# 3. What is a Missing Value?

# When a value is not available for a particular cell, it is called a missing value.
```

```python
# | Name  | Age |  Salary |
# | ----- | --: | ------: |
# | Aman  |  25 |   40000 |
# | Priya |  27 | **NaN** |
# | Rahul |  30 |   50000 |


# Here Priya's salary is missing.

# In Pandas, missing numerical data is commonly represented as:

# NaN

# Meaning:

# Not a Number / missing numerical value
```

```python
# Create a Dataset with Missing Values
import pandas as pd
import numpy as np

data = {
    "Name": ["Aman", "Priya", "Rahul", "Sneha", "Karan"],
    "Age": [25, 27,np.nan, 24, 30],
    "Salary": [40000, np.nan,50000, 45000,np.nan],
    "Department": ["IT",np.nan, "HR", "IT", "Finance"]
}

df = pd.DataFrame(data)

print(df)

# Detect Missing Values
# isnull()
df.isnull()

# df.isna()

# # For normal Pandas EDA purposes: can be used to detect missing values.

# # Count Missing Values
df.isnull().sum()
```

```python
# Missing Value Percentage

# Counting missing values is good. But percentage is often more informative.
# Formula: Missing % =Missing Values / Total Rows × 100

# Python: 
missing_percentage = (df.isnull().sum() / len(df)) * 100

print(missing_percentage)

# This tells us how serious the missing-data problem is.
```

```python
# Why Percentage Matters?
# Suppose:Dataset A :100 rows

# Salary missing: 5

# Missing percentage: 5%
# Dataset B:100 rows

# Salary missing:80

# Missing percentage:80%

# Obviously, both situations should not be handled in exactly the same way.
```

```python
# Find Rows Having Missing Values Sometimes we want to see the actual problematic rows.

print(df.isnull().any(axis=0))
# What does this mean?
# df.isnull() -checks missing values.

# .any(axis=1) means:Check whether any column in each row contains True.

# So we get only rows containing at least one missing value.
```

```python
# Find Rows Where One Specific Column is Missing
# For example, Salary:

df[df["Salary"].isnull()]

# This gives employees whose salary is missing.

# Similarly:

df[df["Age"].isnull()]
```

```python
# # What Do We Do With Missing Values?

# Remove
# Mean
# Median
# Mode
# Forward fill
# Backward fill
# Advanced methods

```

```python
# Method 1 — Remove Missing Rows df.dropna()
print(df)
# This removes rows containing missing values.
clean_df = df.dropna()
# print(clean_df)

#  Problem
# Suppose you have: 100,000 rows and 30,000 rows contain missing values. 
# Removing all of them means losing 30% of your data.
# So don't blindly use:  dropna()

# Remove Rows Only for a Specific Column
# Suppose Salary is extremely important:

df.dropna(subset=["Salary"])

# This removes only rows where Salary is missing.
```

```python
# Method 2 — Fill with Mean

# Suppose:
# Salary
# 40000
# 50000
# 45000
# NaN

# Mean: 45000

# We can fill missing salary with mean:

df["Salary"] = df["Salary"].fillna(df["Salary"].mean())
# But be careful! Mean is sensitive to extreme values.

# Example:

# 30000
# 40000
# 45000
# 500000
# NaN

# The 500000 can heavily affect the mean.
print(df)


df["Age"]=df["Age"].fillna(df["Age"].mean())
print(df["Age"])
```

```python
# Method 3 — Fill with Median

df["Age"] = df["Age"].fillna(df["Age"].median())
print(df)
```

```python
# Method 4 — Mode

# Mode means:

# Most frequently occurring value.

# Example:
# IT
# HR
# IT
# IT
# Finance
# NaN

# Mode:#IT
# We can use:
df["Department"] = df["Department"].fillna(
    df["Department"].mode()
)
print(df)

# This is commonly useful for categorical columns.
```

```python
# | Column/Data                  | Possible method                          |
# | ---------------------------- | ---------------------------------------- |
# | Numerical, relatively normal | Mean                                     |
# | Numerical, skewed/outliers   | Median                                   |
# | Categorical                  | Mode                                     |
# | Very few missing rows        | Drop rows may be reasonable              |
# | Time-series data             | Forward/backward fill may be appropriate |
```

```python
# Part 1 — Univariate Analysis

# What is Univariate Analysis?
# Uni = One, Variate = Variable.

# Univariate analysis studies one variable independently to understand its:
# - Distribution
# - Central tendency
# - Spread or variability
# - Frequency
# - Unusual values

# There are two main types of variables to analyze.-> numerical and categorical

#1.  Numerical
# Age, Salary, Experience
# - Mean
# - Median
# - Standard deviation
# - Histogram
# - Box plot


# 2. Categorical
# Department, City, Product type
# - Unique categories
# - Frequency
# - Mode
# - Bar chart
# - Count plot
```

```python
# # Numerical analysis using Pandas

# # Let's analyze the Salary column.
# # A. Mean
# # df["Salary"].mean()

# # # B. Median
# # df["Salary"].median()

# # The median is the middle value after sorting the observations.
# # - Odd number of observations → middle value.
# # - Even number of observations → average of the two middle values.

# # C. Mode
# df["Salary"].mode()

# # # D. Minimum and maximum
# print("Minimum:", df["Salary"].min())
# print("Maximum:", df["Salary"].max())
# print("Range:", df["Salary"].max() - df["Salary"].min())

# # # The range is the difference between the highest and lowest values.

# # # E. Variance and standard deviation
# print("Variance:", df["Salary"].var())
# print("Standard Deviation:", df["Salary"].std())



# # #Variance batati hai ki values mean se kitni squared distance par spread hain.
# # # Standard deviation spread ko original unit, yani rupees, mein express karti hai.
# # # Standard deviation zyada hone ka matlab ho sakta hai salaries mein variability zyada hai. Iska interpretation data ke context par depend karta hai.


# # # F. Quartiles aur IQR
# # # - (Q1) = 25th percentile
# # # - (Q2) = 50th percentile, ya median
# # # - (Q3) = 75th percentile
# # # IQR middle 50% data ka spread measure karta hai.

# # # IQR = Q3 - Q1


q1 = df["Salary"].quantile(0.25)
q2 = df["Salary"].quantile(0.50)
q3 = df["Salary"].quantile(0.75)

iqr = q3 - q1

print("Q1:", q1)
print("Median:", q2)
print("Q3:", q3)
print("IQR:", iqr)
# # # IQR measures the spread of the middle 50% of observations
# # # and is also used for outlier detection.

# # # G. Get all basic statistics together
print(df["Salary"].describe())



# # # This returns count, mean, standard deviation, minimum, quartiles and maximum.
```

```python
# A. Experience vs Salary
# Question: Kya experience badhne ke saath salary bhi generally badhti hai?
sns.scatterplot(data=df, x="Experience", y="Salary")
plt.title("Experience vs Salary")
plt.xlabel("Years of Experience")
plt.ylabel("Salary")
plt.show()



# Har dot ek employee ko represent karta hai.
# - X-axis par experience hai.
# - Y-axis par salary hai.
# Agar points generally bottom-left se top-right ki taraf jaate hain, toh positive association ka indication mil sakta hai.
# Lekin graph dekhkar ye prove nahi hota ki experience hi salary badhne ka cause hai.


# Correlation calculate karna
print(df["Experience"].corr(df["Salary"]))



# Pearson correlation ki range -1 se +1 hoti hai.
# Value	                            Meaning
# +1 ke paas	         Strong positive linear relationship
# 0 ke paas              	Weak linear relationship
# -1 ke paas           	 Strong negative linear relationship


# Zero ke paas correlation hone ka matlab ye nahi ki koi nonlinear relationship bilkul nahi hai.
```

```python
# B. Department vs Salary
# Question: Kaunse department ki average salary highest hai?
department_salary = df.groupby("Department")["Salary"].mean()
print(department_salary)


# - groupby("Department") — departments ke hisaab se groups banata hai.
# - ["Salary"] — salary column select karta hai.
# - .mean() — har group ki average salary calculate karta hai.

# Multiple statistics ek saath:
summary = df.groupby("Department")["Salary"].agg(   
    ["count", "mean", "median", "min", "max"]
    )
print(summary)



# Ye department-wise salary ka detailed summary dega.
```

```python
# Categorical vs Categorical — Crosstab
# Ab ek Promotion column add karte hain.
df["Promotion"] = [    "Yes", "No", "Yes", "No", "Yes", "Yes",   
                    "No", "No", "Yes", "Yes", "No", "Yes"
                   ]
table = pd.crosstab(df["Department"], df["Promotion"])
print(table)

# Crosstab har department mein Yes aur No promotions ki count show karta hai.


# Heatmap banane ke liye:
sns.heatmap(table, annot=True, fmt="d", cmap="Blues")
plt.title("Department vs Promotion")
plt.show()



# annot=True cells ke andar values show karta hai aur fmt="d" integer formatting use karta hai.
# Fair comparison ke liye kabhi-kabhi counts ke bajaye department-wise percentages dekhna better hota hai.
```

```python
# Hypothesis Testing
# Hypothesis Testing ek statistical method hai jisse hum sample data ke basis par 
# kisi claim ke against evidence evaluate karte hain.


# Real-life example: 
# Ek food delivery company claim karti hai ki uski average delivery time 30 minutes hai.
# Hum sample orders ka data lekar check karna chahte hain ki 
# average delivery time 30 minutes se different hone ka evidence hai ya nahi.
```

```python
# # two important hypothesis

# Null Hypothesis — (H0)
# Ye default claim hota hai. 
# Example: population ki average delivery time 30 minutes hai.
# H0: u = 30


# Alternative Hypothesis —(H1) or (Ha)
# Ye woh claim hai jiske support mein hum evidence check karte hain.
# H1:u != 30

# here u = mean
```

```python
# P-value aur significance level
# - Significance level (alpha): Test se pehle choose kiya gaya threshold, commonly 0.05.
# - P-value: Null hypothesis aur test assumptions ke under,
#  observed result jitna ya usse zyada extreme result milne ki probability.

# Decision rule:
# Condition	           Decision
# p-value ≤ 0.05	       Reject H0
# p-value > 0.05	    Fail to reject H0
```

```python
# Test 1 — Independent t-test
# Independent t-test do independent groups ke population means compare karne ke liye use hota hai.
# Example: Kya Restaurant A aur Restaurant B ke average delivery times mein statistically significant difference hai?
```

```python
from scipy import stats

restaurant_A = [28, 30, 32, 27, 29, 31, 30, 28]
restaurant_B = [35, 38, 33, 36, 40, 34, 37, 39]

t_stat, p_value = stats.ttest_ind(
    restaurant_A,
    restaurant_B,
    equal_var=False
)

print("T-statistic:", t_stat)
print("P-value:", p_value)

if p_value > 0.05:
    print("Significant difference found")
else:
    print("Insufficient evidence of a difference")
```

```python
# stats.ttest_ind() do independent samples ka t-test karta hai.

# - equal_var=False Welch's t-test select karta hai, jo equal population variances assume nahi karta.

# - t_stat test statistic hai.

# - p_value null hypothesis ke against evidence assess karne mein help karta hai.

# - if statement p-value ko significance level 0.05 se compare karta hai.
```

```python
# Test 2 — Chi-square Test
# Chi-square test categorical data ke counts ke beech association check karne ke liye use hota hai.
# Example: Kya payment method aur customer type ke beech association hai?
```

```python
import pandas as pd
from scipy.stats import chi2_contingency

table = pd.DataFrame(
    [[30, 20],
     [15, 35]],
    index=["New Customer", "Returning Customer"],
    columns=["UPI", "Card"]
)

print(table)

chi2, p_value, dof, expected = chi2_contingency(table)

print("Chi-square statistic:", chi2)
print("P-value:", p_value)
print("Degrees of freedom:", dof)
print("Expected frequencies:")
print(expected)

if p_value < 0.05:
    print("Evidence of an association")
else:
    print("Insufficient evidence of an association")
```

```python
# - table observed counts ki contingency table hai.

# - chi2_contingency() observed counts ko null hypothesis ke under expected counts se compare karta hai.

# - dof degrees of freedom hai.

# - expected woh counts hain jo independence assume karne par expected hote.
```

```python
# Yahan null hypothesis hai 
# ki customer type aur payment method independent hain.
```

```python
# Test 3 — ANOVA
# ANOVA ka full form Analysis of Variance hai. Iska use generally teen ya usse zyada independent groups ke population means compare karne ke liye hota hai.
# Example: Kya teen restaurants ke average delivery times different hain?
```

```python
# Hypotheses
# -(H0): Sabhi groups ke population means equal hain.
# - (H1): Kam se kam ek population mean different hai.
```

```python
from scipy import stats

restaurant_A = [28, 30, 29, 31, 27]
restaurant_B = [35, 37, 34, 36, 38]
restaurant_C = [29, 30, 32, 28, 31]

f_stat, p_value = stats.f_oneway(
    restaurant_A,
    restaurant_B,
    restaurant_C
)

print("F-statistic:", f_stat)
print("P-value:", p_value)

if p_value < 0.05:
    print("At least one group mean differs")
else:
    print("Insufficient evidence of a difference")
```

```python
# stats.f_oneway() one-way ANOVA perform karta hai.

# f_stat between-group variation ko within-group variation ke comparison mein represent karta hai.

# Small p-value indicate karti hai ki sabhi population means equal hone ke against evidence hai.
```

```python
# ANOVA significant aane par ye automatically nahi batata ki exactly kaunse two groups different hain. Uske liye post-hoc tests, jaise Tukey HSD, ki zaroorat ho sakti hai.
```

```python
# Situation	                                                                
# Do independent groups ke numerical means compare karne hain-> 	Independent t-test
# Do categorical variables ka association check karna hai-> 	    Chi-square test
# Teen ya zyada independent groups ke means compare karne hain->  One-way ANOVA
```

```python
# 1. Feature Creation

# Feature kya hota hai?
# Feature dataset ka ek input variable ya column hota hai,
# jise hum analysis ya Machine Learning mein use karte hain.

# 
# ex- Age	Salary	Experience	Purchased
# Agar hum predict karna chahte hain ki customer product purchase karega ya nahi:
# - Age — feature
# - Salary — feature
# - Experience — feature
# - Purchased — target variable

# Feature Creation kya hai?
# Existing columns ko use karke ek naya meaningful column banana Feature Creation kehlata hai.

# Real-life examples:
# - Price × Quantity → Total Amount
# - Date of Birth → Age
# - Login Time → Hour of Login
# - Total Marks / Maximum Marks × 100 → Percentage
# - Income − Expenses → Savings
# Naya feature tab useful hai jab woh problem ke baare mein additional information represent karta ho.
```

```python
import pandas as pd

data = {
    "Price": [100, 200, 150, 300],
    "Quantity": [2, 3, 4, 2],
    "Discount_Percent": [10, 5, 0, 20]
}

df = pd.DataFrame(data)

df["Gross_Amount"] = df["Price"] * df["Quantity"]

df["Discount_Amount"] = (
    df["Gross_Amount"] * df["Discount_Percent"] / 100
)

df["Final_Amount"] = (
    df["Gross_Amount"] - df["Discount_Amount"]
)

print(df)
```

```python
# Date aur time se feature creation
df = pd.DataFrame({
    "Order_Date": ["2026-01-05", "2026-02-12", "2026-03-20"]
})

df["Order_Date"] = pd.to_datetime(df["Order_Date"])

df["Year"] = df["Order_Date"].dt.year
df["Month"] = df["Order_Date"].dt.month
df["Day"] = df["Order_Date"].dt.day
df["Day_Name"] = df["Order_Date"].dt.day_name()

print(df)

# Use cases: Monthly sales prediction, weekend shopping analysis aur seasonal demand analysis.
```

```python
# # 2. Binning
# Binning mein numerical values ko groups ya intervals mein divide kiya jata hai.

# Maan lo students ki ages hain:

# 12, 15, 19, 24, 35, 48, 62

# Hum in ages ko categories mein convert kar sakte hain:
# - 0–17 → Minor
# - 18–35 → Young Adult
# - 36–59 → Middle Age
# - 60+ → Senior

# Binning ka main purpose numerical values ko simpler groups mein represent karna hai.
# Binning ke types
# - Equal-width binning: Har interval ki width same hoti hai.
# - Equal-frequency binning: Har group mein approximately same number of observations hoti hain.
# - Domain-based binning: Business rules ya expert knowledge se groups define hote hain.
```

```python
# pd.cut() — Equal-width ya custom intervals
import pandas as pd

ages = [12, 15, 19, 24, 35, 48, 62]

bins= [0, 18, 36, 60, 100]
labels = ["Minor", "Young Adult", "Middle Age", "Senior"]

age_groups = pd.cut(
    ages,
    bins=bins,
    labels=labels,
    right=False
)

print(age_groups)

# ages original numerical data hai.
# bins interval boundaries define karta hai.
# labels har interval ka category name deta hai.
# right=False intervals ko left-inclusive aur right-exclusive banata hai.

# in this example intervals [0, 18), [18, 36), [36, 60) aur [60, 100) hain.
#  Isliye age 18 ko Young Adult aur age 60 ko Senior assign kiya jayega.
```

```python
# pd.qcut() — Equal-frequency binning
import pandas as pd

marks = [20, 35, 40, 45, 50, 60, 70, 80, 90]

groups = pd.qcut(
    marks,
    q=3,
    labels=["Low", "Medium", "High"]
)

print(groups)

# q=3 data ko 3 quantile-based groups mein divide karne ki koshish karta hai.
# labels groups ko names deta hai.
# Har group mein approximately similar number of observations aa sakti hain.
```

```python
# 3. Encoding
# Machine Learning algorithms ko input data ke liye numerical representation chahiye hoti hai. 
# Encoding categorical values ko numerical form mein convert karne ki technique hai.

# encoding ke methods:
# 1. Label Encoding
# 2. Ordinal Encoding
# 3. One-Hot Encoding

# Label Encoding
# Label encoding categorical labels ko integer codes assign karta hai.

from sklearn.preprocessing import LabelEncoder

colors = ["Red", "Blue", "Green", "Red"]

encoder = LabelEncoder()
encoded_colors = encoder.fit_transform(colors)

print(encoded_colors)
print(encoder.classes_)
# LabelEncoder categories ko numerical codes assign karta hai.
#  Mapping ko encoder.classes_ se inspect kar sakte hain.
```

```python
# Ordinal Encoding
# Jab categories ka natural order ho, tab ordinal encoding use kar sakte hain.

# Example:

# Low < Medium < High

from sklearn.preprocessing import OrdinalEncoder

data = [["Low"], ["High"], ["Medium"], ["Low"]]

encoder = OrdinalEncoder(
    categories=[["Low", "Medium", "High"]]
)

encoded = encoder.fit_transform(data)

print(encoded)

# Yahan humne explicitly category order define kiya hai:
# - Low → 0
# - Medium → 1
# - High → 2
# Use cases: Education levels, satisfaction ratings, priority levels — jab categories ka meaningful order ho.
```

```python
# One-Hot Encoding
# Jab categories ka natural order nahi hota, toh one-hot encoding useful hai.
import pandas as pd

df = pd.DataFrame({
    "Color": ["Red", "Blue", "Green", "Red"]
})

encoded_df = pd.get_dummies(
    df,
    columns=["Color"],
    dtype=int
)

print(encoded_df)

# pd.get_dummies() categorical column ke liye separate indicator columns create karta hai.
# columns=["Color"] batata hai ki kis column ko encode karna hai.
# dtype=int output ko 0 aur 1 ke integer form mein rakhta hai.
```

```python
# Data type	                         Suitable method
# Unordered categories	             One-Hot Encoding
# Ordered categories	                 Ordinal Encoding
# Target labels	              LabelEncoder commonly used for classification targets
```

```python
# 4. Feature Scaling -> Feature Scaling mein numerical columns ki values ko ek suitable scale par convert kiya jata hai.

# Example:
# Age	Salary
# 22	25,000
# 30	50,000
# 40	90,000
# Age -> tens 
# Salary -> thousands

# Kuch Machine Learning algorithms distance ya feature magnitudes ke liye sensitive hote hain
# . Scaling un algorithms ke liye important ho sakti hai.
```

```python
# Min-Max Scaling-> Min-Max Scaling values ko generally 0 aur 1 ke beech map karti hai.
# formula -> X(scaled)=     X-Xmin
            #          __________________
            #               Xmax-Xmin

# Example: values [10, 20,30]
# - Minimum = 10
# - Maximum = 30
# - Value 20 ka scaled result = (20 − 10) / (30 − 10) = 0.5
```

```python
from sklearn.preprocessing import MinMaxScaler
import pandas as pd

df = pd.DataFrame({
    "Age": [20, 30, 40, 50],
    "Salary": [20000, 40000, 60000, 80000]
})

scaler = MinMaxScaler()

scaled_values = scaler.fit_transform(df)

scaled_df = pd.DataFrame(
    scaled_values,
    columns=df.columns
)

print(scaled_df)

# MinMaxScaler() scaler object banata hai.
# fit_transform() training data se minimum aur maximum learn karta hai aur values transform karta hai.
# Result ko DataFrame mein convert karke readable output milta hai.
```

```python
# # Standardization — StandardScaler
# # Standardization features ko approximately mean 0 aur standard deviation 1 par transform karti hai.
# # formula-> Z=X-mew
#     #    ______________
#     #          sigma

# # where-> X = original value
# mu = training data ka mean
# sigma = training data ka standard deviation
```

```python
from sklearn.preprocessing import StandardScaler
print(df)
scaler = StandardScaler()

scaled_values = scaler.fit_transform(df)

print(scaled_values)

# StandardScaler har feature ko uske mean aur 
# standard deviation ke basis par transform karta hai.
```

```python
# Robust Scaling-> Agar dataset mein extreme outliers hon,
#  toh RobustScaler useful ho sakta hai.
from sklearn.preprocessing import RobustScaler

scaler = RobustScaler()

scaled_values = scaler.fit_transform(df)

print(scaled_values)

# RobustScaler median aur interquartile range (IQR) use karta hai,
#  isliye extreme values ka effect StandardScaler ke comparison mein generally kam hota hai.
```

```python
# Scaling methods comparison
# Method	      |        Main idea	           |                when to use?
# MinMaxScaler	  |   Values ko fixed range,       |        Jab bounded scale useful ho
#                 |  usually 0–1, mein map karna   |	
# StandardScaler  |	Mean 0, standard deviation 1   |	  KNN, SVM, Logistic Regression aur 
#                 |                                |    PCA jaise methods mein often useful
# RobustScaler	  |   Median aur IQR use karna	   |          Jab outliers present hon
```

```python
# 5. Feature Transformation
# what is transformation/-> Feature Transformation mein numerical values ki mathematical representation 
# ya distribution ko change kiya jata hai.

# Scaling aur transformation related concepts hain, but not same
# - Scaling values ke scale ko adjust karti hai.
# - Transformation values ke pattern ya representation ko change kar sakti hai.

# Example: Income data mein kuch logon ki income ₹20,000 hai aur kuch ki ₹10,00,000.
#  Aisi data distribution right-skewed ho sakti hai.
```

```python
# Log Transformation
# Log transformation large values ke differences ko compress kar sakti hai aur positive, 
# right-skewed data ko less skewed bana sakti hai.


import numpy as np
import pandas as pd

df = pd.DataFrame({
    "Income": [20000, 30000, 50000, 100000, 1000000]
})

df["Log_Income"] = np.log1p(df["Income"])

print(df)

# np.log1p(x) mathematically (log(1+x)) calculate karta hai.
# Yeh zero values ke liye bhi defined hai.
# Log_Income ek transformed feature hai.
```

```python
# Power Transformation
# Scikit-learn mein PowerTransformer data ki distribution ko zyada Gaussian-like banane mein help kar sakta hai.

from sklearn.preprocessing import PowerTransformer

transformer = PowerTransformer(method="yeo-johnson")

transformed_values = transformer.fit_transform(df[["Income"]])

print(transformed_values)
```

```python
# Variance Threshold practical
# Low-variance features ko remove


import pandas as pd
from sklearn.feature_selection import VarianceThreshold

X = pd.DataFrame({
    "Study_Hours": [1, 2, 3, 4, 5],
    "Attendance": [60, 70, 80, 90, 100],
    "Constant_Feature": [1, 1, 1, 1, 1]
})

selector = VarianceThreshold(threshold=0)

selected_values = selector.fit_transform(X)
print(selected_values)

selected_columns = X.columns[selector.get_support()]

print(selected_columns.tolist())

# Explanation:
# - X input features ka DataFrame hai.
# - VarianceThreshold(threshold=0) zero-variance columns ko remove karta hai.
# - fit_transform() feature selection apply karta hai.
# - get_support() batata hai ki kaunse columns retain hue.
```

```python
# 7. Feature Extraction
# Feature Extraction mein original data se ek naya representation banaya jata hai. 
# Ismein multiple original features combine ho sakte hain
#  ya unhe lower-dimensional representation mein convert kiya ja sakta hai.
```

```python
# Example:
# Maan lo dataset mein 10 correlated numerical columns hain. 
# Hum PCA se unhe fewer principal components mein represent kar sakte hain.


# PCA — Principal Component Analysis
# PCA ek dimensionality reduction technique hai. 
# Yeh original numerical features ke linear combinations banata hai, jinhe principal components kehte hain.
```

```python
# PCA ke main goals:
# - Features ki count reduce karna.
# - Data mein maximum possible variance capture karne ki koshish karna.
# - Correlated numerical features ko compact representation mein convert karna.
# - High-dimensional data ko visualize karne mein help karna.
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
import pandas as pd

X = pd.DataFrame({
    "Maths": [70, 80, 90, 60, 85],
    "Science": [75, 82, 95, 65, 88],
    "English": [65, 78, 85, 70, 80]
})

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

pca = PCA(n_components=2)

X_pca = pca.fit_transform(X_scaled)

print(X_pca)
print(pca.explained_variance_ratio_)

# StandardScaler() features ko standardize karta hai.
# fit_transform(X) har feature ke scale ko adjust karta hai.
# PCA(n_components=2) do principal components create karne ko kehta hai.
# pca.fit_transform(X_scaled) standardized data ko 2-dimensional representation mein convert karta hai.
# explained_variance_ratio_ har retained component ke through explain ki gayi variance ka proportion batata hai.
```

```python
# here 3 original columns hain aur 2 extracted components hain.


# Important: PCA components original features jaise directly interpretable nahi hote. 
# Iske liye scaling often important hoti hai, especially jab features different units mein hon.
```

```python
# Feature Selection vs Feature Extraction
# Feature Selection	                                                                  Feature Extraction
# Original columns mein se subset choose karta hai.	                 Original data se new representations banata hai.
# Selected columns original features hote hain.	        Extracted features combinations ya transformed representations ho sakte hain.
# Example: Relevant columns select karna.	                                    Example: PCA components create karna.
# Interpretability often zyada rehti hai.	                                     Interpretability reduce ho sakti hai.
```
