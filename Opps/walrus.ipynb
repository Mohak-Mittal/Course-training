```python
# Walrus Operator := ko Assignment Expression Operator kehte hain.
# it is used to assign a value inside an expression and to use use that value immedietly
```

```python
# Normal assignment vs Walrus Operator
# normal
# name = input("Enter your name: ")

# if name:
#     print("Hello", name)

# Walrus Operator
if ( name := input("Enter your name: ")):
    print("Hello", name)    

# Difference: Walrus Operator input ki value assign bhi karta hai 
# aur condition mein use bhi karta hai.    
```

```python
# Normal approach:
number = int(input("Enter a number: "))
if number > 10:    
    print(number, "is greater than 10")



# Walrus Operator:
if (number := int(input("Enter a number: ")))>10:
    print(number, "is greater than 10")
```

```python
# Suppose user se numbers lete rehna hai jab tak user 0 enter na kare.
# while num!=0:
#     print(num)
while (number := int(input("Enter number (0 to stop): "))) != 0:    
    print("You entered:", number)
```

```python
# multithreading

# Definition: Multithreading is a technique in which multiple threads execute within the same process, 
# allowing a program to handle multiple tasks concurrently.

# suppose your programm is doing 3 tasks:
# - Task 1: File download 
# - Task 2: Data read 
# - Task 3: Email send 

# Normal execution mein tasks ek ke baad ek ho sakte hain.
# Multithreading mein tasks ki progress overlap kar sakti hai,
# especially jab tasks I/O ka wait kar rahe hon.
```

```python
#         Process                                                       	               Thread
# Running program ka instance                                        	Process ke andar execution ka path
# Apni memory space hoti hai	                                    Same process ke threads memory share karte hain
# Generally creation overhead zyada                    	                      Generally lightweight
# Processes ke beech communication comparatively complex	        Shared data communicate karna easy, lekin synchronization zaroori
```

```python
# hpow to make thread in python

# # normally
# def task():
#     print("Task is running")

# task()
# now samw function in thread
import threading

def task():
    print("Task is running")

t1 = threading.Thread(target=task)

t1.start()


# import threading — threading module import kiya.
# def task() — function define kiya.
# threading.Thread(target=task) — thread object banaya.
# t1.start() — thread ka execution start kiya.
```

```python
import threading

def task1():
    print("Task 1 started")

def task2():
    print("Task 2 started")

t1 = threading.Thread(target=task1)
t2 = threading.Thread(target=task2)

t1.start()
t2.start()

t1.join()
t2.join()

print("All tasks completed")

# join() means
# current thread wait kare jab tak specified thread finish na ho jaye.
```

```python
# # start() vs run() vs join()

# Method	                                Meaning
# start()	            Naye thread mein execution start karta hai
# run()	            Thread ka target function execute karta hai; ise directly call karna new thread start nahi karta
# join()	            Calling thread ko wait karwata hai jab tak target thread finish na ho
```

```python
# t1.start()  # New thread starts
t1.join()   # Wait for t1 to finish
```

```python
import threading
import time

def download():
    print("Downloading started")
    time.sleep(2)
    print("Downloading completed")

def process_data():
    print("Processing started")
    time.sleep(2)
    print("Processing completed")

def send_email():
    print("Sending email started")
    time.sleep(2)
    print("Email sent")

t1 = threading.Thread(target=download)
t2 = threading.Thread(target=process_data)
t3 = threading.Thread(target=send_email)

t1.start()
t2.start()
t3.start()

t1.join()
t2.join()
t3.join()

print("All work completed")


# in each task wait of 2 seconds are there

# Sequentially execute karne par approximately 6 seconds lag sakte hain, excluding overhead.
# Threads se ye waits overlap kar sakte hain, isliye total time approximately 2 seconds ke aas-paas ho sakta hai, excluding overhead.
```

```python
# Advanced Multithreading
# sleep() and thread names

# time.sleep() current thread ko kuch time ke liye pause karta hai.

import threading
import time

def task():
    print(threading.current_thread().name)
    time.sleep(2)
    print("Task completed")

t1 = threading.Thread(target=task, name="Worker-1")
t2 = threading.Thread(target=task, name="Worker-2")

t1.start()
t2.start()

t1.join()
t2.join()

print("Main thread completed")

# - current_thread() current executing thread ka object deta hai.
# - name thread ka naam deta hai.
# - sleep(2) current thread ko approximately 2 seconds wait karwata hai.
# - join() thread complete hone ka wait karta hai.
```

```python
# Race Condition
# Race condition tab hoti hai jab multiple threads shared data ko access ya modify karte hain
# aur final result execution ke timing/order par depend karne lagta hai.
```

```python
# Example: bank account mein ₹1,000 hain. Do threads same time par ₹800 withdraw karne ki koshish karte hain.
# Agar dono threads balance ko ₹1,000 read kar lein, toh dono withdrawal allow kar sakte hain. Isse incorrect balance ya transaction handling ho sakti hai.
```

```python
import threading

counter = 0

def increment():
    global counter

    for _ in  range(100):
        counter += 1

t1 = threading.Thread(target=increment)
t2 = threading.Thread(target=increment)

t1.start()
t2.start()

t1.join()
t2.join()

print("Counter:", counter)


# Lekin shared state par unsynchronized read-modify-write operations se race conditions ho sakti hain.
# Solution: Lock
```

```python
# lock->Lock ensure karta hai ki ek waqt mein sirf ek thread protected critical section execute kare.
import threading

counter = 0
lock = threading.Lock()

def increment():
    global counter

    for _ in range(10000):
        with lock:
            counter += 1

t1 = threading.Thread(target=increment)
t2 = threading.Thread(target=increment)

t1.start()
t2.start()

t1.join()
t2.join()

print("Counter:", counter)
```

```python
# Thread 1 requests lock
#          |
# Critical section: counter += 1
#          |
# Lock released automatically

# Agar doosra thread wait kar raha hai, lock available hone par woh aage badh sakta hai.

# advantage ye hai ki block se exit hote hi lock release ho jata hai, exception aane par bhi.
```

```python
# ThreadPoolExecutor — modern approach
# Jab bahut saare tasks chalane hon, har task ke liye manually thread banana inconvenient ho sakta hai. 
# ThreadPoolExecutor threads ka pool manage karta hai.



from concurrent.futures import ThreadPoolExecutor
import time

def download_file(file_id):
    print("Downloading file", file_id)
    time.sleep(1)
    return f"File {file_id} downloaded"

with ThreadPoolExecutor(max_workers=3) as executor:
    results = executor.map(download_file, [1, 2, 3, 4, 5])

    for result in results:
        print(result)


# maximum 3 worker threads ek waqt mein pool ke through tasks handle karenge. Baaki tasks workers free hone par execute honge.
# - max_workers=3 — pool mein maximum 3 workers.
# - executor.map() — function ko multiple inputs par run karta hai.
# - with — pool ka lifecycle manage karta hai.
# - results — results input order mein deta hai, chahe tasks ka completion order alag ho.        
```

```python
# Concept	                                                Purpose
# sleep()	                                Current thread ko temporarily pause karna
# current_thread()                         	Current thread ki information
# Race condition	                    Shared data par timing-dependent incorrect behavior
# Lock	                                    Critical section ko protect karna
# ThreadPoolExecutor	                       Worker threads ka pool manage karna
# join()	                                    Thread completion ka wait karna
```
