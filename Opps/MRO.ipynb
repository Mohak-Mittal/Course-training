```python
# MRO — Method Resolution Order
# “Agar multiple classes mein same method hai, Python sabse pehle kis class mein method dhundhega?”
```

```python
class A:
    def show(self):
        print("A")


class B(A):
    pass


class C(B):
    def show(self):
        print("C")


obj = B()
obj.show()

# Python method ko search karega:

# C → B → A → object

# C mein show() mil gaya → wahi execute.

# Is order ko MRO kehte hain.
```

```python
# Python classes ko ek particular order mein search karta hai 
# jab kisi attribute ya method ko access kiya jata hai.
```

```python
# Method 1 — .mro()
print(C.mro())
# C → B → A → object
```

```python
# Method 2 — __mro__
print(B.__mro__)
#Python mein ultimately classes ka connection object se hota hai.

# C
# ↓
# B
# ↓
# A
# ↓
# object

# object Python ki fundamental base class hai. Isliye MRO ke end mein generally:object hoga
```

```python
class A:
    pass


class B:
    def show(self):
        print("B")


class C(A, B):
    pass


obj = C()

obj.show()

# Why?

# C ke parents hain:  C(A, B)
#  Python search karega: C → A → B → object 
# C mein show() nahi hai.
# A mein mil gaya.

```

```python
print(C.mro())
```

```python
# # Diamond Problem
#        A
#       / \
#      B   C
#       \ /
#        D


class A:
    def show(self):
        print("A")


class B(A):
    pass


class C(A):
    def show(self):
        print("C")


class D(B, C):
    pass
obj = D()
obj.show()

# Why? D ka MRO: D → B → C → A → object
# Python pehle D dekhta hai. D mein show() nahi.
# Then B: B.show() mil gaya. So B execute.
```

```python
print(D.mro())
```
