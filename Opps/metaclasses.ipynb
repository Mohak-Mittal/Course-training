```python
# now actual Metaclass
class MyMeta(type):
    pass
class Student(metaclass=MyMeta):
    pass

print(type(Student))
```

```python
# actual use
class MyMeta(type):

    def __new__(cls, name, bases, attrs):
        
        print("Creating class:", name)

        return super().__new__(
            cls,
            name,
            bases,
            attrs
        )


class Student(metaclass=MyMeta):
    pass
```
