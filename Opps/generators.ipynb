```python
# Definition: A generator is a special type of iterator that produces values one at a time
# , only when needed.

# in simple-> Generator ek aisa mechanism hai jo saari values ek saath dene ke bajaye ek-ek karke value produce karta hai.

# for example 1 se 5 tak numbers print
numbers = [1, 2, 3, 4, 5]

print(numbers)
# here all values are in list 
# Generator bhi ye values de sakta hai, lekin values ko ek-ek karke produce karta hai.
```

```python
# Normal function vs Generator function

# normal-
def get_numbers():
    return [1, 2, 3, 4, 5]


result = get_numbers()

print(result)

# generator function

def get_numbers():
    yield 1
    # yield 2
    # yield 3
    # yield 4
    # yield 5


result = get_numbers()

print(result)

# get_numbers() ab ek generator object return karta hai. Ye abhi saari values produce nahi karta.
```

```python

def get_numbers():
   for i in range(1,8):
    yield(i)





for numbers in get_numbers():
        print(numbers)
```

```python
# yield 

#  important keyword of generators is  yield.
# features
# Ek value produce karta hai.
# Function ki current state ko preserve karta hai.
# Generator ko temporarily pause karta hai.
# Next request par wahi se execution resume hota hai.
```

```python
# Example:

def demo():
    print("First")
    yield 10

    print("Second")
    yield 20

    print("Third")
    yield 30


g = demo()

print(next(g))
print(next(g))
print(next(g))
```

```python
# __set__() — value set karna

class AgeDescriptor:

    def __set__(self, instance, value):

        if value >= 0:
            instance._age = value
        else:
            raise ValueError("Age cannot be negative")


class Student:

    age = AgeDescriptor()

    def __init__(self, age):
        self.age = age


s1 = Student(20)

print(s1.age)           

# problem 

# Humne __get__() nahi banaya.

# So:

# print(s1.age)

# properly value return nahi karega.
```

```python
# __get__() — value read karna
class AgeDescriptor:

    def __get__(self, instance, owner):

        return instance._age

    def __set__(self, instance, value):

        if value >= 0:
            instance._age = value
        else:
            raise ValueError("Age cannot be negative")

class Student:

    age = AgeDescriptor()

    def __init__(self, age):
        self.age = age


s1 = Student(20)

print(s1.age)        




# s1.age = 20

# internally descriptor ko activate karti hai.

# Conceptually:

# s1.age = 20
#        ↓
# AgeDescriptor.__set__()
#        ↓
# value validation
#        ↓
# s1._age = 20


# print(s1.age)

# Conceptually:

# s1.age
#   ↓
# AgeDescriptor.__get__()
#   ↓
# s1._age
#   ↓
# 20
```
