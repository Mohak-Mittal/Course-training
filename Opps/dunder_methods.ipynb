```python
# # Magic / Dunder Methods
# Python mein kuch methods already special purpose ke liye defined hote hain.
# Inke naam double underscore __ se start aur end hote hain.
```

```python
# Dunder = Double Underscore

# Examples:

# __init__
# __str__
# __len__
# __add__
# __eq__
```

```python
# 1. __init__()
class Student:

    def __init__(self, name):
        self.name = name
s1 = Student("Rahul")    
print(s1)    
```

```python
# # 2. __str__()
# # object ko print karne par readable output dene ke liye use hota hai.

# # Without __str__():

# class Student:

#     def __init__(self, name):
#         self.name = name


# s1 = Student("Rahul")

# print(s1)


# # with __str__():

class Student:

    def __init__(self, name):
        self.name = name

    def __str__(self):
        return self.name


s1 = Student("Rahul")

print(s1)
```

```python
# 3. __len__()
# Object par len() ka behaviour define

class Team:

    def __init__(self, players):
        self.players = players

    def __len__(self):
        return len(self.players)


team = Team(["Rahul", "Aman", "Riya", "Neha","abc"])

print(len(team))
```

```python
# 4. __add__()-> + operator ka behaviour define


class Number:

    def __init__(self, value):
        self.value = value

    def __add__(self, other):
        return self.value + other.value


n1 = Number(10)
n2 = Number(20)

print(n1 + n2)
```

```python
# 5. __sub__()

class Number:

    def __init__(self, value):
        self.value = value

    def __sub__(self, other):
        return self.value - other.value


n1 = Number(30)
n2 = Number(10)

print(n1 - n2)
```

```python
# 6. __eq__()->== operator ke liye.

class Student:

    def __init__(self, roll_no):
        self.roll_no = roll_no

    def __eq__(self, other):
        return self.roll_no == other.roll_no


s1 = Student(101)
s2 = Student(10)


print(s1 == s2)
```

```python
# 7. __lt__()-> < operator ke liye.

class Student:

    def __init__(self, marks):
        self.marks = marks

    def __lt__(self, other):
        return self.marks < other.marks


s1 = Student(70)
s2 = Student(80)

print(s1 < s2)
```

```python
# 8. __gt__()-> > operator ke liye.

class Student:

    def __init__(self, marks):
        self.marks = marks

    def __gt__(self, other):
        return self.marks > other.marks


s1 = Student(90)
s2 = Student(80)

print(s1 > s2)
```

```python
# Operator Overloading
# Operator Overloading means giving a special meaning to an operator when it is used with objects of our class.

# 7i+7j
# 6+9i
class ComplexNumber:

    def __init__(self, real, imaginary):
        self.real = real
        self.imaginary = imaginary

    def __add__(self, other):
        real_part = self.real + other.real
        imaginary_part = self.imaginary + other.imaginary

        return ComplexNumber(real_part, imaginary_part)

    def __str__(self):
        return f"{self.real} + {self.imaginary}i"


c1 = ComplexNumber(2, 3)
c2 = ComplexNumber(4, 5)

c3 = c1 + c2

print(c3)
```
