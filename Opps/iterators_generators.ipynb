```python
# iter()-> iterable ko iterator  me convert

num = [10,20,60,77]
it = iter(num)
print(it)
```

```python
# next(): will return next value


num = [10,20,60,77]
it = iter(num)
it1 = next(it)
print(it1)

```

```python
# StopIteration()
try:
    num = [10,20,60,77]
    it = iter(num)
    print(next(it))
    print(next(it))
    print(next(it))
    print(next(it))
    print(next(it))
except Exception as StopIteration :
    print(StopIteration )                           



```

```python
# generators: a special function that uses yield and return value one by one

# why do we use it->

def numbers():
    return list(range(1000000000))
```

```python
# return: only one return will execute
def test():
    return 10

    return 56
```

```python
# yield: all at one time will execute

def test():
    yield 10

    yield 56

    yield 58 
```

```python
def number():
      yield 10

      yield 56

      yield 58 

n1 = number()
print(next(n1))
print(next(n1))
print(next(n1))
```

```python
#  generator with loop
def num():
    for i in range(1,6):
        yield i
        
for i in num():
    print(i)        
```

```python
# square generator
def sq():
    for i in range(1,6):
        yield i*i
        
for i in sq():
    print(i)        

```
