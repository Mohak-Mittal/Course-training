```python
# MULTILEVEL 

# grandparent -> parent -> child

class Grandparent:
    def show(self):
        print("heloo am your grandpa")
class Parent(Grandparent):
    def abc(self):
        print("am your father")

class Child(Parent):
    pass

a1 = Parent()
a1.show()

a2 = Child()
a2.abc()
a2.show()
```

```python
# Method Overriding: Definition

# Method overriding occurs when a child class provides its own implementation of a method that already exists in the parent class.
class Animal:

    def sound(self):
        print("Animal makes sound")
class Dog(Animal):

    def sound(self):
        super().sound()
        print("Dog barks")
d1 = Dog()

d1.sound()                
```

```python
# # without encapsulation:

# class Bank:
#     def __init__(self):
#         self.balance = 10000000

# b1 = Bank()
# print(b1.balance)    
# b1.balance = 5
# print(b1.balance)    



# # with encapsulation:
class Bank:
    def __init__(self):
        self.__balance = 10000
    def get_balance(self) :
        return self.__balance

b1 = Bank()
b1.__balance=800
print(b1.get_balance()) 
 
```

```python
# getter method: for data read

class Bank:
    def __init__(self):
        self.__balance = 10000
    def get_balance(self):
        return self.__balance

b1 = Bank()
print(b1.get_balance())        

```

```python
# setter: private method ko update krne ke liye

class Bank:
    def __init__(self):
        self.__balance = 10000
    def set_balance(self,amount):
        self.__balance = amount
    def get_balance(self):
        return self.__balance 

b1 = Bank()
b1.set_balance(200000)
print(b1.get_balance())    
```
