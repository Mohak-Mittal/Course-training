```python
# Duck Typing means that an object's behavior is more important than its actual type or class.
# Object kis class ka hai, usse zyada important hai ki object kya kar sakta hai.
```

```python
# Suppose:
class Dog:
    def bark(self):
        print("Bark")
class Cat:
    def speak(self):
        print("Meow")

dog = Dog()
cat = Cat()

# Traditional thinking:# dog → Dog # cat → Cat

# But in Python 

def make_sound(animal):
    animal.speak()

make_sound(dog)
make_sound(cat)    

# Function ko ye check karne ki zarurat nahi hai:

# if type(animal) == Dog:
# or
# if type(animal) == Cat:

# Usse sirf ek cheez matter karti hai:
# animal.speak()

# Agar object ke paas speak() method hai, kaam ho jayega.

# This is Duck Typing.
```

```python
class Dog:

    def speak(self):
        print("Dog says: Bark")


class Cat:

    def speak(self):
        print("Cat says: Meow")


class Human:

    def speak(self):
        print("Human says: Hello")
def make_speak(obj):
    obj.speak()
dog = Dog()

cat = Cat()
human = Human()

make_speak(dog)
make_speak(cat)
make_speak(human)      
# Dog, Cat, aur Human ka koi common parent nahi hai. why?
#  make_speak() teeno ke saath kaam kar raha hai.
# Why? Because teeno objects ke paas: speak() method hai.      
```

```python
# | Inheritance Polymorphism                | Duck Typing                                |
# | --------------------------------------- | ------------------------------------------ |
# | Parent-child relationship ho sakta hai  | Common parent required nahi                |
# | Type hierarchy important ho sakti hai   | Behaviour important hai                    |
# | Common interface inherited ho sakta hai | Object ke paas required method hona enough |
# | `Dog(Animal)`                           | `Dog` aur `Animal` unrelated ho sakte hain  |
# | More explicit                           | More flexible                              |
```

```python
class UPI:

    def pay345(self):
        print("Payment using UPI")


class Card:

    def pay(self):
        print("Payment using Card")


class Cash:

    def pay(self):
        print("Payment using Cash")



def make_payment(method):
    method.pay()
make_payment(UPI())
make_payment(Card())
make_payment(Cash())    

# Function ko ye matter nahi:

# "Are you UPI?"
# "Are you Card?"
# "Are you Cash?"

# sirf matter karta hai:

# Do you have pay()?

# That's Duck Typing.
```
