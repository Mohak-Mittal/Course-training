```python
class BankAccount:

    def __init__(self, balance):
        self.__balance = balance

    def show_balance(self):
        print(self.__balance)

    def withdraw(self, amount):

        if amount <= self.__balance:
            self.__balance -= amount
            print("Money withdrawn")

        else:
            print("Insufficient balance")
account = BankAccount(8000)

account.withdraw(300)

account.show_balance()            
```

```python
# getter-># Private data ko safely read/access karna.
class Student:

    def __init__(self,name, marks):
        self.name = name
        self.__marks = marks

    def get_marks(self):
        return self.__marks
s1 = Student("Rahul", 85)
# print(s1.self.__marks())
print(s1.get_marks())

# self.__marks is private so print(s1.__marks) is not valid 
# instead s1.get_marks() can be used

# Getter = Get/Read the value
```

```python
# Setter Method
# Private data ko safely modify/update karna.
class Student:

    def __init__(self, name, marks):
        self.name = name
        self.__marks = marks

    def get_marks(self):
        return self.__marks

    def set_marks(self, marks):
        if 0 <= marks <= 100:
            self.__marks = marks
        else:
            print("Invalid marks!")
s1 = Student("Rahul", 85)

print(s1.get_marks())

s1.set_marks(95)

print(s1.get_marks())
s1.set_marks(105)            
```

```python
class BankAccount:

    def __init__(self, account_no, balance):
        self.account_no = account_no
        self.__balance = balance

    # Getter
    def get_balance(self):
        return self.__balance

    # Setter
    def set_balance(self, balance):
        if balance >= 0:
            self.__balance = balance
        else:
            print("Balance cannot be negative!")


account = BankAccount("A101", 10000)

# Getter
print(account.get_balance())

# Setter
account.set_balance(15000)

print(account.get_balance())

# Invalid value
account.set_balance(-500)
```

```python
from abc import ABC, abstractmethod



class Animal(ABC):

    @abstractmethod
    def sound(self):
        pass
class Dog(Animal):

    def sound(self):
        print("Dog barks")
class Cat(Animal):

    def sound(self):
        print("Cat meows")
dog = Dog()
cat = Cat()


dog.sound()
cat.sound()         
animal = Animal()       
```

```python
class Vehicle(ABC):

    @abstractmethod
    def start(self):
        pass
class Car(Vehicle):

    def start(self):
        print("Car starts with key")
class Bike(Vehicle):

    def start(self):
        print("Bike starts with button")
class ElectricCar(Vehicle):

    def start(self):
        print("Electric car starts silently")   
f=Car()
c = Bike()
d = ElectricCar()                         
c.start()
```
