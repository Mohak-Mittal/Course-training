```python
# Definition: Parsing is the process of analyzing input data and converting it into a structured format that a program can understand and process.
```

```python
# for example you have this data:
# data = "Ankita,22,Ludhiana"

# if you want to separate it
data = "Ankita,22,Ludhiana"

parts = data.split(",")

print(parts)

# now all are separated but in strings

name = parts[0]
age = int(parts[1])
city = parts[2]

print(name)
print(age + 1)
print(city)
# input string ko parse karke useful values mein convert kiya.


```

```python
# String Parsing

log = "2026-10-12|ERROR|Database connection failed"
# date, log level aur message alag chahiye.

log = "2026-10-12|ERROR|Database connection failed"

date, level, message = log.split("|")

print("Date:", date)
print("Level:", level)
print("Message:", message)
```

```python
# JSON Parsing
# JavaScript Object Notation

import json

json_data = '{"name": "Ankita", "age": 22, "city": "Ludhiana"}'

student = json.loads(json_data)

print(student)
print(student["name"])
print(student["age"])

# json.loads()            vs json.load()
# Function	               Input
# json.loads()          	JSON string
#                           json.load()	       File object containing JSON
```
